## Ejercicio 1

```python
#1.1 Creo un arreglo para los datos de luminosidad solar
L = np.array([1.0 , 2.5 , 0.8 , 5.0 , 10.0])
#creo un arreglo para los datos de distancias en parsecs
d = np.array([10 , 25 , 15 , 40 , 60])

print("LUMINOSIDADES: ", L)
print("DISTANCIA: ", d)
```

```python
#1.2 Para el flujo observado de una estrella (F) se usa la luminosidad (L)
#d la distancia, 4πd⁴ simboliza el area de una esfera de radio d.

#antes de utilizar la formula se hace unas conversiones de unidades:
def calcular_flujo(L,d):
    Lsolar = 3.828 * 10**26 #la luminosidad solar en watts
    pcs_a_m = 3.0856776 * 10**16 #de parsec a metros

    luminosidad = L * Lsolar
    distancia = d * pcs_a_m

    F = luminosidad / (4 * np.pi * distancia**2)
    return F
F = calcular_flujo(L,d)
print("Flujo en W/m²: ", F)
```

```python
#1.3 Indice de mayor flujo:
indice_mayor_flujo = np.max(F)
print("Indice de mayor flujo: ", indice_mayor_flujo)

#Indice menor flujo:
indice_menor_flujo = np.min(F)
print("Indice de menor flujo: ", indice_menor_flujo)
```

```python
#1.4 promedio y mediana del flujo
flujo_promedio = np.mean(F)
flujo_mediana = np.median(F)

print("Flujo promedio: ", flujo_promedio)
print("Flujo mediano: ", flujo_mediana)
```

```python
#1.5 buscamos valores de flujos que superen el limite de deteccion
limite = 10**(-11)
estrellas_detectadas = F > limite
print("Estrellas detectadas: ", estrellas_detectadas)
```

```python
#1.6 mostramos los indices en donde se encuentran las estrellas detectadas
indices = np.where(estrellas_detectadas)[0] #np.where entrega una estructura que contiene los índices, en el arreglo tomo el primer elemento ([0])
print("Indices detectados: ", indices)
```

```python
#1.7 orden de mayor a menor flujo
orden_flujos = np.argsort(F)[::-1] #np.argsort entrega en orden de menor a mayor, pero debe mostrarse de mayor a menor asi que se usa [::-1] para "voltear" el orden
print("orden de los flujos de mayor a menor: ", orden_flujos)
```

```python
#2.1 arreglo temporal
t = np.linspace(0,10,500)
#use linspace ya que range entrega datos en int, mientras que linspace si puede con 500 mediciones equidistantes entre 0 y 10
```

```python
#2.2 flujo normalizado de una estrella
F = np.ones(500) #np.ones crea un arreglo donde cada elemento se inicia con valor 1
```

```python
#2.3 flujo con ruido gaussiano
ruido_g = rng.normal(0, 0.002, 500) #rng.normal genera números aleatorios con distribuicion normal
F = F + ruido_g
```

```python
#2.4 intervalo de transito
transito = (t >=4.5) & (t <=5.0) 
F[transito] -= 0.02 #disminuir el flujo en 0.02 para los puntos dentro del intervalo del transito
```

```python
#2.5 flujo promedio de luz
flujo_promedio = np.mean(F)
print("Flujo promedio: ", flujo_promedio)
```

```python
#2.6 flujo estandar
desviacion = np.std(F)
print("la deviacion estandar del flujo es: ", desviacion)
```

```python
#2.7  valor minimo
flujo_min = np.min(F)
indice_min = np.argmin(F)
print("el flujo minimo es: ", flujo_min, "y su indice: ", indice_min)
```

```python
#2.8 tiempo del valor minimo
tiempo_min = t[indice_min]
print("tiempo del flujo minimo: ", tiempo_min)
```

```python
#2.9 seleccionar todos los puntos que cumplan una condicion
sigma = np.std(F)
deteccion_puntos = F < (flujo_promedio - 3 *sigma)
print(deteccion_puntos)
```

```python
#2.10 puntos posibles pertenecientes al transito
indices_p = np.where(deteccion_puntos)[0]
print(indices_p)
```

```python
#2.11 profundidad transito
profundidad = 1 - flujo_min
print("Profundidad:", profundidad)
```

```python
#2.12 calcular la razón entre el radio del planeta y el radio estelar
radio = np.sqrt(profundidad)

print("Rp/R* =", radio)
```

```python
#2.13 obtener los tiempos en los que se detecta el tránsito mediante la condicion
tiempo = t[deteccion_puntos]
print("tiempos detecctados: ", tiempo)
print()
#comparo con el intervalo original del tránsito
print("intervalo original: 4.5 a 5.0 años")
print("Intervalo detectado:", tiempo.min(), "a", tiempo.max())
```
